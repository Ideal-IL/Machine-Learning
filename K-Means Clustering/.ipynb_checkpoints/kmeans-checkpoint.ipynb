```python
#These are the dataset given to us from kagglehub 
#https://www.kaggle.com/datasets/zubairdhuddi/shopping-dataset
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.cluster import KMeans
from sklearn.preprocessing import StandardScaler

shopbehavior = pd.read_csv("shopping_behavior_updated.csv")

shopbehavior.head()
```

```python
#Let's start with the K-Means

#Here's our x variables
x = shopbehavior[["Age","Purchase Amount (USD)","Review Rating","Previous Purchases"]]
```

```python
#Let's scale our data
x_scaled = StandardScaler().fit_transform(x)

#StandardScalar() - creates the scaler
#.fit_transforms(x) - learns the scaling and applies it to x
```

```python
#Create the k-means model
model = KMeans(n_clusters=3, random_state=42)

#Fitting the model
model.fit(x_scaled)

#Adding labels
shopbehavior["Cluster"] = model.labels_
3
#n_clusters - makes a cluster of 3
#model.labels_ - gives labels to model
```

```python
#The best way to visualize K-Means is to scatter plot the clusters
plt.scatter(shopbehavior["Age"], shopbehavior["Purchase Amount (USD)"], c = shopbehavior["Cluster"])
plt.xlabel("Age")
plt.ylabel("Purchase Amount (USD)")
plt.title("K-Means Customer Clusters")
plt.show()

#c=shopbehavior["Clusters"] gives a color to each cluster
```

```python
#Lets look at the average for the three clusters we made
shopbehavior.groupby("Cluster")[["Age","Purchase Amount (USD)","Review Rating","Previous Purchases"]].mean()
```

```python
#Cluster 0 - Purple
#Older + higher-spending customers

#Cluster 1 - Teal
#Middle-aged + lower-spending customers

#Cluster 2 - Yellow
#Younger + higher-spending customers
```
